```python
import polars
import matplotlib.pyplot as plt
```

```python
from pyprojroot import here
PROJECT_ROOT = here()
HISTORY_DIR = PROJECT_ROOT / "history"
FIG_DIR = PROJECT_ROOT / "figs" / "history"
FIG_DIR.mkdir(parents=True, exist_ok=True)
```

```python
def plot(history_name):
    df = polars.read_csv(HISTORY_DIR/f"{history_name}.csv")
    plt.plot(df["train_loss"], label="train_loss")
    plt.plot(df["validate_loss"], label="validate_loss")
    plt.xlim(0, 29)
    plt.ylim(0.01, 0.05)
    plt.xlabel("epoch")
    plt.ylabel("loss")
    plt.legend()
    plt.savefig(FIG_DIR/f"{history_name}.png")
    plt.show()
```

```python
plot("freesound_delay_1")
# plot("v2_noise_fullband_delay_1")
plot("noise_fullband_delay_1")
```
