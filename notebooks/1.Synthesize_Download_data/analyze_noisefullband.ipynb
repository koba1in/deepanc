```python
from pyprojroot import here
PROJECT_ROOT = here()
DATASET_DIR = PROJECT_ROOT / "noise_fullband" / "train"
DATASET_SAVE_DIR = PROJECT_ROOT / "noise_fullband" / "resampled"
DATASET_SAVE_DIR.mkdir(parents=True, exist_ok=True)

RESAMPLE_RATE = 16000
DATASET_STAT_DIR = PROJECT_ROOT / "noise_fullband" / "statistics"
DATASET_STAT_DIR.mkdir(parents=True, exist_ok=True)
stat_file = DATASET_STAT_DIR / "train.csv"
stat_file.touch(exist_ok=True)
```

```python
import csv
import soundfile as sf
from construct_dataset_from_freesound.resample import resampler
metadata = sf.info(PROJECT_ROOT/"noise_fullband/train/breath_spit_Freesound_validated_26803_0.wav")
print(metadata)
```

```python

from concurrent.futures import ThreadPoolExecutor

def get_file_info(wave_file):
    if wave_file.is_file() and wave_file.suffix.lower() in [".wav", ".flac"]:
        try:
            metadata = sf.info(wave_file)
            return [wave_file.name, metadata.frames]
        except Exception:
            return None
    return None

def resample_file(wave_file,):
    if wave_file.is_file() and wave_file.suffix.lower() in [".wav", ".flac"]:
        try: 
            waveform, sample_rate = sf.read(wave_file, dtype="float32", )
            waveform = resampler(waveform, sample_rate, RESAMPLE_RATE)
            sf.write(DATASET_SAVE_DIR/wave_file.name, waveform, RESAMPLE_RATE, subtype="PCM_16", )
            return None
        except Exception:
            return None
    return None
```

```python
wave_files = list(DATASET_SAVE_DIR.iterdir())
with ThreadPoolExecutor() as executor:
    results = [res for res in executor.map(get_file_info, wave_files) if res]

with open(stat_file, mode="w", newline="", encoding="utf-8") as csvfile:

    writer = csv.writer(csvfile, )
    writer.writerow(["filename", "frames"])
    writer.writerows(results)
```

```python
import polars as pl
import numpy as np

sampling_rate = RESAMPLE_RATE
df = pl.read_csv(stat_file)
frames = df["frames"].to_numpy()
durations = frames / sampling_rate
print(f"総録音時間：{durations.sum()/60/60:.2f}時間")
```
